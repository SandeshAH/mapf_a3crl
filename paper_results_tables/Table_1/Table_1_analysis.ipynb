# Training Performance Comparison on Medium Complexity Environments

This notebook generates **Table 1** for the paper.

Each model has its own evaluation experiments and result files.  
For each model we compute the **average success rate across its evaluation experiments**.

Models will be processed sequentially in this notebook.

Current models:
1. Imitation Learning (IL)
2. Reinforcement Learning (RL Only) Model
3. Hybrid Standard Imitation Learning Model
4. Hybrid Coordinate Model
5. PRIMAL Original Model

```python
import re
import pandas as pd
from pathlib import Path
```

Function to Parse Success Rate

```python
def extract_success_rates(file_path):

    with open(file_path, "r") as f:
        text = f.read()

    success_rates = re.findall(r"Success Rate:\s*([\d.]+)%", text)

    success_rates = [float(rate) for rate in success_rates]

    return success_rates
```

Imitation Learning (IL) Model Results

```python
model_name = "Imitation Learning (IL)"

results_file = Path(
    "Results_table_1/Eval_EXP5_and_EXP7_IL_only_results.txt"
)

success_rates = extract_success_rates(results_file)

average_success = sum(success_rates) / len(success_rates)

print("Model:", model_name)
print("Success Rates:", success_rates)
print("Average Success Rate:", round(average_success,2), "%")
```

```python
il_result = {
    "Model": model_name,
    "Average Success Rate (%)": round(average_success,2)
}

il_result
```

```python
table1_results = []

table1_results.append(il_result)

df_table1 = pd.DataFrame(table1_results)

df_table1
```
